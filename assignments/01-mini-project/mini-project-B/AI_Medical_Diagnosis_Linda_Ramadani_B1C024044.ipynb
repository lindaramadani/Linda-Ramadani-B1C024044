# **Analisis dan Jawaban Pertanyaan**

### **1. Analisis Algoritma Mana yang Lebih Baik**
*(Catatan: Analisis ini didasarkan pada hasil eksekusi umum dataset Breast Cancer Wisconsin)*

*   **Akurasi & ROC-AUC:** Bandingkan nilai akurasi dari output sel di atas. Umumnya, kedua model memberikan performa yang sangat tinggi (di atas 94%). Namun, kita perlu melihat metrik krusial lainnya.
*   **False Negative (Sangat Penting):** Perhatikan sel kiri-bawah pada Confusion Matrix masing-masing model (aktual *Malignant* tetapi diprediksi *Benign*). Model yang memiliki nilai False Negative (FN) lebih kecil adalah model yang lebih aman untuk digunakan dalam konteks medis.

### **2. Mengapa Evaluasi Model Tidak Cukup Hanya Menggunakan Accuracy?**

Dalam konteks medis (seperti klasifikasi kanker payudara):

1.  **Dampak Fatal dari Kesalahan Klasifikasi (False Negative vs False Positive):**
    *   **False Positive (FP):** Pasien sehat didiagnosis kanker. Ini menyebabkan kepanikan psikologis dan pemeriksaan medis lanjutan, namun akhirnya diketahui salah.
    *   **False Negative (FN):** Pasien penderita kanker ganas didiagnosis sehat/jinak sehingga tidak mendapatkan penanganan medis. **Kesalahan ini sangat fatal karena dapat mengancam nyawa pasien.**
2.  **Keterbatasan Akurasi:**
    *   Akurasi hanya menghitung persentase kebenaran prediksi secara global tanpa membedakan tipe kesalahan.
    *   Model dengan akurasi 98% terlihat bagus, tetapi jika sisa 2% kesalahan tersebut adalah penderita kanker ganas yang tidak terdeteksi (FN), model tersebut sangat berbahaya jika diimplementasikan.
3.  **Pentingnya Recall (Sensitivity):**
    *   Dalam diagnosis kanker, kita lebih mengutamakan nilai **Recall** yang tinggi untuk kelas *Malignant* (ganas). Recall mengukur seberapa banyak kasus positif asli yang berhasil diidentifikasi oleh model, meminimalisir nilai False Negative mendekati nol.
